```python
"""
Company Employee Dataset Cleaning
"""

import pandas as pd
import numpy as np

df = pd.read_csv('Day11_Messy_Company_Employee_Dataset.csv')

print("Initial Shape:", df.shape)
print("\nMissing Values:")
print(df.isnull().sum())
print("\nDuplicates:", df.duplicated().sum())
```

```python
# Clean column names
df.columns = df.columns.str.strip()
```

```python
# Create copy
df_cleaned = df.copy()
```

```python
# Fill missing numeric values
df_cleaned['Age'] = df_cleaned['Age'].fillna(df_cleaned['Age'].median())
df_cleaned['Annual_Salary'] = df_cleaned['Annual_Salary'].fillna(df_cleaned['Annual_Salary'].mean())
df_cleaned['Experience_Years'] = df_cleaned['Experience_Years'].fillna(df_cleaned['Experience_Years'].mean())
df_cleaned['Performance_Score'] = df_cleaned['Performance_Score'].fillna(df_cleaned['Performance_Score'].mean())
```

```python
# Fill missing categorical values
df_cleaned['Department'] = df_cleaned['Department'].fillna(df_cleaned['Department'].mode()[0])
df_cleaned['Gender'] = df_cleaned['Gender'].fillna(df_cleaned['Gender'].mode()[0])
df_cleaned['Work_Mode'] = df_cleaned['Work_Mode'].fillna(df_cleaned['Work_Mode'].mode()[0])
df_cleaned['City'] = df_cleaned['City'].fillna('Unknown')
df_cleaned['Job_Title'] = df_cleaned['Job_Title'].fillna('Unknown')
```

```python
# Fix inconsistent gender
df_cleaned['Gender'] = df_cleaned['Gender'].replace({
    'M': 'Male', 'm': 'Male', 'male': 'Male', 'MALE': 'Male',
    'F': 'Female', 'f': 'Female', 'female': 'Female', 'FEMALE': 'Female',
    'N/A': 'Other'
})
```

```python
# Fix department
df_cleaned['Department'] = df_cleaned['Department'].str.strip().str.title()
df_cleaned['Department'] = df_cleaned['Department'].replace({
    'Engineer': 'Engineering',
    'engineering': 'Engineering'
})
```

```python
# Fix work mode
df_cleaned['Work_Mode'] = df_cleaned['Work_Mode'].str.upper().str.strip()
df_cleaned['Work_Mode'] = df_cleaned['Work_Mode'].replace({
    'OFFICE': 'Office', 'REMOTE': 'Remote', 'HYBRID': 'Hybrid'
})

# Fix city
df_cleaned['City'] = df_cleaned['City'].str.strip().str.title()
df_cleaned['City'] = df_cleaned['City'].replace({'N/A': 'Unknown', 'Delhi ': 'Delhi'})

# Fix salary (remove commas)
df_cleaned['Annual_Salary'] = df_cleaned['Annual_Salary'].astype(str).str.replace(',', '').astype(float)
```

```python
# Convert data types
df_cleaned['Age'] = df_cleaned['Age'].astype(int)
df_cleaned['Joining_Date'] = pd.to_datetime(df_cleaned['Joining_Date'], errors='coerce')

# Remove duplicates
df_cleaned = df_cleaned.drop_duplicates()
```

```python
print("\nCleaned Shape:", df_cleaned.shape)
print("\nMissing Values After Cleaning:")
print(df_cleaned.isnull().sum())
```

```python
print("\nComparison:")
print(f"Rows: {df.shape[0]} -> {df_cleaned.shape[0]}")
print(f"Missing: {df.isnull().sum().sum()} -> {df_cleaned.isnull().sum().sum()}")
print(f"Duplicates: {df.duplicated().sum()} -> {df_cleaned.duplicated().sum()}")
```

```python
# Export
df_cleaned.to_csv('Cleaned_Employee_Dataset.csv', index=False)
print("\nSaved as 'Cleaned_Employee_Dataset.csv'")
```
