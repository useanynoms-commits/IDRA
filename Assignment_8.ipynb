```python
"""
E-commerce Sales Data Analysis
"""

import pandas as pd
import numpy as np

df = pd.read_csv('Day8_Ecommerce_Sales_Dataset.csv')
# Basic info
print(f"\nShape: {df.shape}")
print("\nFirst 5 rows:")
print(df.head())
```

```python
# Statistics
print(df[['Quantity','Unit_Price','Discount_Percent','Total_Sales','Rating']].describe())
print(f"\nTotal Orders: {len(df)}")
print(f"Total Revenue: Rs.{df['Total_Sales'].sum():,.2f}")
print(f"Average Order: Rs.{df['Total_Sales'].mean():,.2f}")
```

```python
# Category analysis
cat = df.groupby('Category').agg({'Total_Sales':['sum','mean','min','max'], 'Quantity':'sum', 'Order_ID':'count'})
cat.columns = ['Revenue','Avg','Min','Max','Quantity','Orders']
print(cat.sort_values('Revenue', ascending=False))
```

```python
# City analysis
city = df.groupby('City')['Total_Sales'].sum().sort_values(ascending=False)
print(city.head(10))
```

```python
# Product analysis
prod = df.groupby('Product')['Total_Sales'].sum().sort_values(ascending=False)
print(prod.head(10))
```

```python
# Payment method analysis
pay = df.groupby('Payment_Method').agg({'Total_Sales':'sum', 'Order_ID':'count'})
pay.columns = ['Revenue','Orders']
print(pay.sort_values('Revenue', ascending=False))
```

```python
# Region analysis
reg = df.groupby('Region').agg({'Total_Sales':'sum', 'Order_ID':'count', 'Quantity':'sum'})
reg.columns = ['Revenue','Orders','Quantity']
print(reg.sort_values('Revenue', ascending=False))
```

```python
# Filtering operations

high_orders = df[df['Total_Sales'] > 10000]
print(f"High value orders (>Rs.10,000): {len(high_orders)}")
print(high_orders[['Order_ID','Customer','Product','Total_Sales']].head())

high_discount = df[df['Discount_Percent'] > 10]
print(f"\nOrders with >10% discount: {len(high_discount)}")
print(f"Avg sales after discount: Rs.{high_discount['Total_Sales'].mean():.2f}")

top_rated = df[df['Rating'] >= 4.8]
print(f"\nTop rated orders (>=4.8): {len(top_rated)}")
```

```python
# Customer analysis
cust = df.groupby('Customer').agg({'Total_Sales':'sum', 'Order_ID':'count'})
cust.columns = ['Total_Spent','Orders']
print(cust.sort_values('Total_Spent', ascending=False).head(5))
```

```python
# Discount impact
discount_ranges = pd.cut(df['Discount_Percent'], bins=[0,5,10,15,100], labels=['0-5%','6-10%','11-15%','>15%'])
disc = df.groupby(discount_ranges).agg({'Total_Sales':['count','mean','sum']})
disc.columns = ['Orders','Avg_Value','Revenue']
print(disc)
```

```python
# Observations
print(f"""
1. Total Orders: {len(df)}
2. Total Revenue: Rs.{df['Total_Sales'].sum():,.2f}
3. Best Category: {cat['Revenue'].idxmax()} (Rs.{cat['Revenue'].max():,.2f})
4. Best Product: {prod.idxmax()} (Rs.{prod.max():,.2f})
5. Top City: {city.idxmax()} (Rs.{city.max():,.2f})
6. Top Customer: {cust['Total_Spent'].idxmax()} (Rs.{cust['Total_Spent'].max():,.2f})
7. Most Used Payment: {pay['Orders'].idxmax()}
8. Best Region: {reg['Revenue'].idxmax()}
9. Avg Discount: {df['Discount_Percent'].mean():.2f}%
10. Avg Rating: {df['Rating'].mean():.2f}
""")
```
