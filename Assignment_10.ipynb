```python
"""
Flight Operations Data Analysis
"""
import pandas as pd
import numpy as np

df = pd.read_csv('Day10_Flight_Operations_Dataset.csv')

print("Dataset Shape:", df.shape)
print("\nFirst 5 rows:")
print(df.head())

print("\nDataset Info:")
df.info()

print("\nBasic Statistics:")
print(df.describe())

print("\nNull values:")
print(df.isnull().sum())
```

```python
# Data cleaning
df['Delay_Minutes'] = df['Delay_Minutes'].fillna(0)
df['Passenger_Satisfaction'] = df['Passenger_Satisfaction'].fillna(df['Passenger_Satisfaction'].mean())
df['Flight_Date'] = pd.to_datetime(df['Flight_Date'])
df['Month'] = df['Flight_Date'].dt.month

print("\nMissing values after cleaning:")
print(df.isnull().sum())
```

```python
# Filtering
delayed = df[df['Flight_Status'] == 'Delayed']
on_time = df[df['Flight_Status'] == 'On Time']
cancelled = df[df['Flight_Status'] == 'Cancelled']
high_delay = df[df['Delay_Minutes'] > 60]

print(f"\nDelayed: {len(delayed)}, On Time: {len(on_time)}, Cancelled: {len(cancelled)}")
print(f"Flights with >60 min delay: {len(high_delay)}")
```

```python
# Sorting
print("\nTop 10 Longest Delays:")
print(df.sort_values('Delay_Minutes', ascending=False)[['Flight_ID', 'Airline', 'Origin', 'Destination', 'Delay_Minutes']].head(10))

print("\nTop 10 Highest Satisfaction:")
print(df.sort_values('Passenger_Satisfaction', ascending=False)[['Flight_ID', 'Airline', 'Passenger_Satisfaction']].head(10))
```

```python
# Grouping and aggregation
airline_stats = df.groupby('Airline').agg({
    'Passengers': 'sum',
    'Flight_ID': 'count',
    'Delay_Minutes': 'mean',
    'Passenger_Satisfaction': 'mean'
}).round(2)
airline_stats.columns = ['Total_Passengers', 'Total_Flights', 'Avg_Delay', 'Avg_Satisfaction']
print("\nAirline Performance:")
print(airline_stats.sort_values('Avg_Satisfaction', ascending=False))

class_stats = df.groupby('Travel_Class').agg({
    'Passengers': 'sum',
    'Flight_ID': 'count',
    'Average_Ticket_Price': 'mean',
    'Passenger_Satisfaction': 'mean'
}).round(2)
class_stats.columns = ['Total_Passengers', 'Total_Flights', 'Avg_Price', 'Avg_Satisfaction']
print("\nTravel Class Analysis:")
print(class_stats)

weather_impact = df.groupby('Weather').agg({
    'Flight_ID': 'count',
    'Delay_Minutes': 'mean',
    'Passenger_Satisfaction': 'mean'
}).round(2)
weather_impact.columns = ['Total_Flights', 'Avg_Delay', 'Avg_Satisfaction']
print("\nWeather Impact:")
print(weather_impact)

booking_stats = df.groupby('Booking_Channel').agg({
    'Flight_ID': 'count',
    'Average_Ticket_Price': 'mean',
    'Passenger_Satisfaction': 'mean'
}).round(2)
booking_stats.columns = ['Total_Flights', 'Avg_Price', 'Avg_Satisfaction']
print("\nBooking Channel Analysis:")
print(booking_stats)

meal_stats = df.groupby('Meal_Preference').agg({
    'Flight_ID': 'count',
    'Passenger_Satisfaction': 'mean'
}).round(2)
meal_stats.columns = ['Total_Flights', 'Avg_Satisfaction']
print("\nMeal Preference Analysis:")
print(meal_stats)
```

```python
# Observations
print(f"""
1. Total flights: {len(df)}
2. Airlines: {df['Airline'].nunique()}
3. Avg delay: {df['Delay_Minutes'].mean():.2f} minutes
4. Avg satisfaction: {df['Passenger_Satisfaction'].mean():.2f}/5
5. Best airline: {airline_stats['Avg_Satisfaction'].idxmax()} ({airline_stats['Avg_Satisfaction'].max():.2f})
6. Worst airline: {airline_stats['Avg_Satisfaction'].idxmin()} ({airline_stats['Avg_Satisfaction'].min():.2f})
7. Most delayed weather: {weather_impact['Avg_Delay'].idxmax()} ({weather_impact['Avg_Delay'].max():.2f} min)
8. Best meal: {meal_stats['Avg_Satisfaction'].idxmax()} ({meal_stats['Avg_Satisfaction'].max():.2f})
9. Most common class: {class_stats['Total_Flights'].idxmax()}
10. Busiest origin: {df.groupby('Origin')['Flight_ID'].count().idxmax()}
11. Avg load factor: {df['Load_Factor'].mean():.2f}%
12. Delayed flights: {len(delayed)} ({len(delayed)/len(df)*100:.1f}%)
13. On-time flights: {len(on_time)} ({len(on_time)/len(df)*100:.1f}%)
14. Cancelled flights: {len(cancelled)} ({len(cancelled)/len(df)*100:.1f}%)
15. Top booking channel: {booking_stats['Total_Flights'].idxmax()}
""")
```
