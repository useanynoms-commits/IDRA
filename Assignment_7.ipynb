```python
import pandas as pd
import numpy as np

# Load the CSV dataset
df = pd.read_csv('Day7_Student_Dataset.csv')
print("="*60)
print("DATASET SUMMARY REPORT")
print("="*60)

# Dataset shape
print(f"\nDataset Shape: {df.shape[0]} rows and {df.shape[1]} columns")

# Column names
print(f"\nColumns: {list(df.columns)}")

# First few rows
print("\nFirst 5 rows:")
print(df.head())

# Last few rows
print("\nLast 5 rows:")
print(df.tail())

# Random samples
print("\nRandom 5 samples:")
print(df.sample(5))
```

```python
# 2. Dataset Information
print("\n" + "="*60)
print("DATASET INFORMATION")
print("="*60)

# Data types and memory usage
print("\nData Types:")
print(df.dtypes)

# Info summary
print("\nInfo Summary:")
df.info()
```

```python
# 3. Descriptive Statistics
print("\n" + "="*60)
print("DESCRIPTIVE STATISTICS")
print("="*60)

# Statistical summary for numerical columns
print("\nNumerical Columns Statistics:")
print(df.describe())

# Statistical summary for categorical columns
print("\nCategorical Columns Statistics:")
print(df.describe(include=['object']))
```

```python
# 4. Missing Values Analysis
print("\n" + "="*60)
print("MISSING VALUES ANALYSIS")
print("="*60)

# Count of missing values per column
print("\nMissing Values per Column:")
print(df.isnull().sum())

# Percentage of missing values
print("\nPercentage of Missing Values:")
print((df.isnull().sum() / len(df)) * 100)
```

```python
# 5. Additional Exploratory Analysis
print("\n" + "="*60)
print("ADDITIONAL ANALYSIS")
print("="*60)

# Unique values count
print("\nUnique Values Count:")
print(df.nunique())

# Value counts for categorical columns
print("\nCity Distribution:")
print(df['City'].value_counts())

print("\nCourse Distribution:")
print(df['Course'].value_counts())
```

```python
# Summary statistics
print("\n" + "="*60)
print("SUMMARY STATISTICS")
print("="*60)

print(f"Total Records: {len(df)}")
print(f"Total Features: {len(df.columns)}")
print(f"Missing Values: {df.isnull().sum().sum()}")
print(f"Duplicate Rows: {df.duplicated().sum()}")
print(f"Memory Usage: {df.memory_usage().sum() / 1024:.2f} KB")
```
